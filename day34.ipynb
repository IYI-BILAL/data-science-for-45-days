```python
import nltk
nltk.download('stopwords')


from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
text = "My name is Mohammed Bilal Shaikh"

words = word_tokenize(text)
print(words)

stop_words = set(stopwords.words('english'))
print(stop_words)

# filter data
filtered = []

for word in words:
  if word.lower() not in stop_words:
    filtered.append(word)

    # print filtered data
    print('filtered data')
    print(filtered)
```

```python
# stemming -> converts words into their root form

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

print(stemmer.stem("playing"))
print(stemmer.stem("played"))
print(stemmer.stem("player"))
```

```python
# create a project with combination of word_tokenize,stopwords and PorterStemmer

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

text = "My name is Mohammed Bilal Shaikh and I love to do coding in pyhton"

# Tokenization
tokens = word_tokenize(text)

# Stopwords
stop_words = set(stopwords.words("english"))
filtered = [word for word in tokens if word.lower() not in stop_words]


# Stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(word) for word in filtered]

print(text)
print(tokens)
print(filtered)
print(stemmed)
```
